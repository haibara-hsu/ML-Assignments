```python
# Logistic Regression
from sklearn.linear_model import LogisticRegression
x = df.iloc[:,2:10].values
y = df.iloc[:,1].values
print(x.shape)
print(y.shape)
print(df.dtypes)
```

```python
clf = LogisticRegression()
clf.fit(x,y)
```

```python
print ('Coefficients: ', clf.coef_)
print ('Intercept: ', clf.intercept_)
```

```python
print(clf.predict([[61370680,0,3,0,6,403174,0,0]]))
print(clf.predict([[48467332,64776,0,841,8,39267,56,1]]))
```

```python
print('score:',clf.score(x,y))
```

```python
#Polynomial regression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
poly_reg = PolynomialFeatures(degree = 2)

x_poly = poly_reg.fit_transform(x)
reg2 = LinearRegression()

reg2.fit(x_poly, y)
```

```python
print(reg2.predict(poly_reg.fit_transform([[61370680,0,3,0,6,403174,0,0]])))
print(reg2.predict(poly_reg.fit_transform([[48467332,64776,0,841,8,39267,56,1]])))
```

```python
print ('Coefficients: ', reg2.coef_)
print ('Intercept: ', reg2.intercept_)
```

```python
print('score:',reg2.score(x_poly,y))
```
